```python
from beir import util
from beir.datasets.data_loader import GenericDataLoader

dataset = "scidocs"

url = f"https://public.ukp.informatik.tu-darmstadt.de/thakur/BEIR/datasets/{dataset}.zip"

data_path = util.download_and_unzip(
    url,
    "datasets"
)

corpus_beir, queries_beir, qrels = GenericDataLoader(
    data_folder=data_path
).load(split="test")
```

```python
import pandas as pd

qrels_records = []
for query_id, doc_scores in qrels.items():
    for doc_id, score in doc_scores.items():
        qrels_records.append({'query_id': query_id, 'doc_id': doc_id, 'score': score})

qrels_df = pd.DataFrame(qrels_records)

print(type(qrels_df))
print(qrels_df.head())
print(qrels_df.shape)
```

```python
print(qrels_df.columns)
```

```python
relevant_per_query = (
    qrels_df.groupby("query_id")
    .size()
)

print(relevant_per_query.describe())
```

```python
print("Number of queries with qrels:", relevant_per_query.shape[0])
print("Total relevance pairs:", len(qrels_df))
print("Average relevant docs per query:", relevant_per_query.mean())
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
print(model)
```

```python
query = queries_df.iloc[0]["text"]

embedding = model.encode(query)

print("Query:", query)
print("Embedding shape:", embedding.shape)
print("First 10 values:", embedding[:10])
```

```python
document = corpus_df.iloc[0]["text"]

doc_embedding = model.encode(document)

print("Document length:", len(document.split()))
print("Embedding shape:", doc_embedding.shape)
```

```python
from sentence_transformers.util import cos_sim

query_embedding = model.encode(
    queries_df.iloc[0]["text"],
    convert_to_tensor=True
)

doc_embeddings = model.encode(
    corpus_df.iloc[:10]["text"].tolist(),
    convert_to_tensor=True
)

similarities = cos_sim(query_embedding, doc_embeddings)

print(similarities)
```

```python
scores = similarities[0].cpu().numpy()

for i, score in enumerate(scores):
    print(i, score)
```

```python
print(corpus_df["text_words"].describe(percentiles=[0.90, 0.95, 0.99]))
```

```python
short_doc = corpus_df.iloc[0]["text"]
medium_doc = corpus_df.iloc[100]["text"]
long_doc = corpus_df.iloc[20000]["text"]

for name, doc in [
    ("short", short_doc),
    ("medium", medium_doc),
    ("long", long_doc)
]:
    print(name, len(doc.split()), "words")
```

```python
embeddings = model.encode(
    [short_doc, medium_doc, long_doc],
    convert_to_tensor=True
)

print(embeddings.shape)
```

```python
texts = corpus_df["text"].tolist()

tokens = model.tokenizer(
    texts,
    truncation=False,
    padding=False
)

token_lengths = [len(x) for x in tokens["input_ids"]]

corpus_df["token_length"] = token_lengths
```

```python
print(
    corpus_df["token_length"].describe(
        percentiles=[0.90, 0.95, 0.99]
    )
)
```

```python
print(
    "Documents over model limit:",
    (corpus_df["token_length"] > model.max_seq_length).sum()
)

print(
    "Percentage over model limit:",
    (corpus_df["token_length"] > model.max_seq_length).mean() * 100
)
```
